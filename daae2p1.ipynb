```python
import random

# Generate DNA sequence
random.seed(42)
dna = ''.join(random.choice("ACGT") for _ in range(50000))
pattern = "ACGTAC"


# Naive Search
def naive_search(text, pattern):
    n = len(text)
    m = len(pattern)
    comparisons = 0
    matches = 0

    for i in range(n - m + 1):
        j = 0
        while j < m:
            comparisons += 1
            if text[i + j] != pattern[j]:
                break
            j += 1
        if j == m:
            matches += 1

    return matches, comparisons


# KMP
def compute_lps(pattern):
    lps = [0] * len(pattern)
    length = 0
    i = 1

    while i < len(pattern):
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        elif length != 0:
            length = lps[length - 1]
        else:
            lps[i] = 0
            i += 1

    return lps


def kmp_search(text, pattern):
    lps = compute_lps(pattern)

    i = j = 0
    comparisons = 0
    matches = 0

    while i < len(text):
        comparisons += 1

        if text[i] == pattern[j]:
            i += 1
            j += 1

            if j == len(pattern):
                matches += 1
                j = lps[j - 1]
        else:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

    return matches, comparisons


# Rabin-Karp
def rabin_karp(text, pattern):
    d = 256
    q = 101

    n = len(text)
    m = len(pattern)

    h = pow(d, m - 1, q)

    p = 0
    t = 0

    comparisons = 0
    matches = 0

    for i in range(m):
        p = (d * p + ord(pattern[i])) % q
        t = (d * t + ord(text[i])) % q

    for s in range(n - m + 1):

        if p == t:
            for j in range(m):
                comparisons += 1
                if text[s + j] != pattern[j]:
                    break
            else:
                matches += 1

        if s < n - m:
            t = (d * (t - ord(text[s]) * h) + ord(text[s + m])) % q
            if t < 0:
                t += q

    return matches, comparisons


naive_match, naive_comp = naive_search(dna, pattern)
kmp_match, kmp_comp = kmp_search(dna, pattern)
rk_match, rk_comp = rabin_karp(dna, pattern)

print("DNA Length :", len(dna))
print("Pattern :", pattern)

print("\nNaive Search")
print("Matches :", naive_match)
print("Comparisons :", naive_comp)

print("\nKMP Search")
print("Matches :", kmp_match)
print("Comparisons :", kmp_comp)

print("\nRabin-Karp Search")
print("Matches :", rk_match)
print("Comparisons :", rk_comp)
```
